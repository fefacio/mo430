```python
import networkit as nk

from src.utils import graph_seed
from src.config import DATA_DIR
from src.converter import nk_to_demds, nk_to_numds
```

```python
n = 10_000_000
beta = 2.7
K=10

g_seed = graph_seed(
    n,
    beta,
    1001,
)


nk.setSeed(g_seed, False)

generator = nk.generators.HyperbolicGenerator(
    n=n,
    k=K,
    gamma=beta,
)



G = generator.generate()


beta_name = str(beta).replace(".", "_")

instance_name = (
    f"n-{n}_"
    f"beta-{beta_name}_"
    f"g-{1001}"
)



instance_dir = DATA_DIR / instance_name
instance_dir.mkdir(parents=True, exist_ok=True)

demds_path = instance_dir / "demds.txt"
numds_path = instance_dir / "numds.txt"

nk_to_demds(G, demds_path)
nk_to_numds(G, numds_path)

del G
del generator
```
