```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import time

import networkit as nk
import pandas as pd

from src.config import DATA_DIR
from src.converter import nk_to_demds, nk_to_numds
from src.runner import run_instance
from src.utils import graph_seed
```

# Generate all graphs

Betas = {2.1, 2.4, 2.7, 3.0, 3.4} <br>
Number of nodes (N) = 10.000.000 <br>
Average node degree (K) = 12 <br>
Seeds = 5 <br>

```python
N_VALUES = [
#    10_000,
#    100_000,
#    1_000_000,
    10_000_000,
]

BETAS = [
    2.1,
    2.4,
    2.7,
    3.0,
    3.4,
]


K = 12

N_GRAPHS = 10


TIME_LIMIT = 10

BASE_GRAPH_SEED = 12345

```

```python

all_results = []

for n in N_VALUES:

    for beta in BETAS:

        for repetition in range(1, N_GRAPHS + 1):

            g_seed = graph_seed(
                n,
                beta,
                repetition,
            )

            print(
                f"\n"
                f"N={n:,} | "
                f"beta={beta} | "
                f"graph={repetition}/{10} | "
                f"seed={g_seed}"
            )


            nk.setSeed(g_seed, False)

            generator = nk.generators.HyperbolicGenerator(
                n=n,
                k=K,
                gamma=beta,
            )

            generation_start = time.perf_counter()

            G = generator.generate()

            generation_time = (
                time.perf_counter()
                - generation_start
            )


            beta_name = str(beta).replace(".", "_")

            instance_name = (
                f"n-{n}_"
                f"beta-{beta_name}_"
                f"g-{repetition}"
            )



            instance_dir = DATA_DIR / instance_name
            instance_dir.mkdir(parents=True, exist_ok=True)

            demds_path = instance_dir / "demds.txt"
            numds_path = instance_dir / "numds.txt"

            nk_to_demds(G, demds_path)
            nk_to_numds(G, numds_path)

            del G
            del generator

```

# Run algorithms

```python
ALGORITHM_SEEDS = [1, 2, 3, 4, 5]

for n in N_VALUES:

    for beta in BETAS:

        for repetition in range(1, N_GRAPHS + 1):
            
            beta_name = str(beta).replace(".", "_")

            instance_name = (
                f"n-{n}_"
                f"beta-{beta_name}_"
                f"g-{repetition}"
            )
            
            instance_dir = DATA_DIR / instance_name

            for algorithm_seed in ALGORITHM_SEEDS:

                print(
                    f"  running seed={algorithm_seed}"
                )

                results = run_instance(
                    instance_dir=instance_dir,
                    beta=beta,
                    seed=algorithm_seed,
                    time_limit=TIME_LIMIT,
                )
                print(results)
```
